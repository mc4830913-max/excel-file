```python
def logic_operations(a, b):
    print("First value:", a)
    print("Second value:", b)

    print("AND (a and b):", a and b)
    print("OR (a or b):", a or b)
    print("NOT (not a):", not a)
    print("NOT (not b):", not b)
logic_operations(10,20)
```
